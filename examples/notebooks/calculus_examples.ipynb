# rStar-Math: Advanced Calculus Examples

This notebook demonstrates how rStar-Math handles complex calculus problems.

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from src.core.mcts import MCTS
from src.core.ppm import ProcessPreferenceModel
from src.models.model_interface import ModelFactory
```

## Setup Components

```python
mcts = MCTS.from_config_file('config/default.json')
ppm = ProcessPreferenceModel.from_config_file('config/default.json')
model = ModelFactory.create_model('openai', os.getenv('OPENAI_API_KEY'), 'config/default.json')
```

## 1. Derivatives and Integration

```python
calculus_problems = [
    "Find the derivative of f(x) = sin(x)cos(x)",
    "Integrate ∫x²e^x dx",
    "Find the second derivative of f(x) = ln(x)",
    "Solve the differential equation dy/dx = x + y"
]

for problem in calculus_problems:
    print(f"Problem: {problem}\n")
    action, trajectory = mcts.search(problem)
    
    print("Solution Steps:")
    for step in trajectory:
        confidence = ppm.evaluate_step(step['state'], model)
        print(f"- {step['state']}")
        print(f"  Confidence: {confidence:.2f}\n")
    print("-" * 50 + "\n")
```

## 2. Limits and Series

```python
limit_problems = [
    "Find the limit of (1 + 1/n)^n as n approaches infinity",
    "Find the sum of the infinite series 1 + 1/2 + 1/4 + 1/8 + ...",
    "Determine if the series Σ(1/n) converges"
]

for problem in limit_problems:
    print(f"Problem: {problem}\n")
    action, trajectory = mcts.search(problem)
    
    print("Solution Steps:")
    for step in trajectory:
        confidence = ppm.evaluate_step(step['state'], model)
        print(f"- {step['state']}")
        print(f"  Confidence: {confidence:.2f}\n")
    print("-" * 50 + "\n")
```

## 3. Visualization of Solutions

Let's visualize some of the calculus concepts.

```python
def plot_function_and_derivative(f_str: str):
    """Plot a function and its derivative."""
    problem = f"Find the derivative of f(x) = {f_str}"
    action, trajectory = mcts.search(problem)
    
    # Extract derivative from solution
    derivative_str = trajectory[-1]['state']
    
    # Create plot
    x = np.linspace(-5, 5, 100)
    
    # Original function
    f = eval(f"lambda x: {f_str}")
    plt.plot(x, f(x), label=f'f(x) = {f_str}')
    
    # Derivative
    try:
        f_prime = eval(f"lambda x: {derivative_str}")
        plt.plot(x, f_prime(x), '--', label=f"f'(x) = {derivative_str}")
    except:
        print("Could not plot derivative")
    
    plt.grid(True)
    plt.legend()
    plt.title(f"Function and its Derivative")
    plt.show()

# Example plots
functions = [
    "x**2",
    "np.sin(x)",
    "np.exp(x)"
]

for f_str in functions:
    plot_function_and_derivative(f_str)
```
