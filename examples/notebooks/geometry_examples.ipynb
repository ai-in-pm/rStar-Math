# rStar-Math: Geometry and Proofs

This notebook demonstrates how rStar-Math handles geometric problems and mathematical proofs.

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Rectangle, Polygon
from src.core.mcts import MCTS
from src.core.ppm import ProcessPreferenceModel
from src.models.model_interface import ModelFactory
```

## Setup Components

```python
mcts = MCTS.from_config_file('config/default.json')
ppm = ProcessPreferenceModel.from_config_file('config/default.json')
model = ModelFactory.create_model('openai', os.getenv('OPENAI_API_KEY'), 'config/default.json')
```

## 1. Basic Geometry Problems

```python
geometry_problems = [
    "Find the area of a circle with radius 5",
    "Calculate the volume of a sphere with radius 3",
    "In a right triangle with legs 3 and 4, find the hypotenuse",
    "Find the area of a regular hexagon with side length 2"
]

for problem in geometry_problems:
    print(f"Problem: {problem}\n")
    action, trajectory = mcts.search(problem)
    
    print("Solution Steps:")
    for step in trajectory:
        confidence = ppm.evaluate_step(step['state'], model)
        print(f"- {step['state']}")
        print(f"  Confidence: {confidence:.2f}\n")
    print("-" * 50 + "\n")
```

## 2. Geometric Proofs

```python
proof_problems = [
    "Prove that the sum of angles in a triangle is 180 degrees",
    "Prove the Pythagorean theorem",
    "Prove that the diagonals of a rectangle are equal"
]

for problem in proof_problems:
    print(f"Problem: {problem}\n")
    action, trajectory = mcts.search(problem)
    
    print("Proof Steps:")
    for step in trajectory:
        confidence = ppm.evaluate_step(step['state'], model)
        print(f"- {step['state']}")
        print(f"  Confidence: {confidence:.2f}\n")
    print("-" * 50 + "\n")
```

## 3. Visualization of Geometric Concepts

```python
def plot_right_triangle():
    """Visualize Pythagorean theorem."""
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # Draw triangle
    ax.plot([0, 3, 0, 0], [0, 0, 4, 0], 'b-', linewidth=2)
    
    # Add squares on each side
    ax.add_patch(Rectangle((0, 0), 3, 3, fill=False, color='red'))
    ax.add_patch(Rectangle((3, 0), 4, 4, fill=False, color='green', angle=90))
    ax.add_patch(Polygon([[0, 0], [0, 4], [-5, 4], [-5, 0]], fill=False, color='blue'))
    
    # Add labels
    ax.text(1.5, -0.5, '3', ha='center')
    ax.text(-0.5, 2, '4', va='center')
    ax.text(1, 2, '5', ha='center', va='center', rotation=-45)
    
    ax.set_aspect('equal')
    ax.grid(True)
    plt.title("Pythagorean Theorem: a² + b² = c²")
    plt.show()

def plot_circle_properties():
    """Visualize circle properties."""
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # Draw circle
    circle = Circle((0, 0), 5, fill=False)
    ax.add_patch(circle)
    
    # Draw radius
    ax.plot([0, 5], [0, 0], 'r-', label='Radius')
    
    # Draw diameter
    ax.plot([-5, 5], [0, 0], 'g--', label='Diameter')
    
    ax.set_aspect('equal')
    ax.grid(True)
    plt.legend()
    plt.title("Circle Properties")
    plt.show()

# Generate visualizations
plot_right_triangle()
plot_circle_properties()
```
